```python
df=spark.read.table("benefit_modernization_dev.bronze.join_patients")
df_appointments=spark.read.table("benefit_modernization_dev.bronze.join_appointments")

spark.conf.set("spark.sql.adaptive.enabled", "false")
spark.conf.set("spark.sql.adaptive.coalescePartitions.enabled", "false")
spark.conf.set("spark.sql.adaptive.skewJoin.enabled", "false")

# ---- Disable broadcast joins ----
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", "-1")

# ---- Fix shuffle behavior ----
spark.conf.set("spark.sql.shuffle.partitions", "200")



new_df=df.join(df_appointments,df.patient_id==df_appointments.patient_id)

new_df.explain()

new_df.write.mode("overwrite").format("noop").save()
```
